## 1.3 Seccion de deteccion de nulos
Se buscan los renglones nulos, se imprimen, se genera un conjunto sin estos valores nulos, se eliminan los espacios exteriores de cada oracion a su vez como pares vacios, sucesivamente se hace una comprobacion para quitar esos pares sin texto. Se calculan las longitudes de cada frase y se añade al data y describimos sus percentiles por cantidad de tokens por frase. Finalmente se muestran ejemplos aleatorios


```python

import pandas as pd

print("Nulos por columna:")
print(df.isna().sum())

# 2. Limpiar los pares incompletos
data = df.dropna(subset=["english", "spanish"]).copy()

# 3. Eliminar espacios exteriores y pares vacíos
for col in ["english", "spanish"]:
    data[col] = data[col].str.strip()

data = data[
    data["english"].ne("") &
    data["spanish"].ne("")
].copy()

# 4. Estadísticas de longitud
data["en_tokens"] = data["english"].str.split().str.len()
data["es_tokens"] = data["spanish"].str.split().str.len()

print("\nLongitud de segmentos:")
print(data[["en_tokens", "es_tokens"]].describe(
    percentiles=[0.25, 0.5, 0.75, 0.95, 0.99]
))

# 5. Duplicados exactos de pares
print("\nPares duplicados:", data.duplicated(
    subset=["english", "spanish"]
).sum())

# 6. Ejemplos aleatorios
print(data.sample(10, random_state=42)[
    ["english", "spanish"]
].to_string(index=False))
```
