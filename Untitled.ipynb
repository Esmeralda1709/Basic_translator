## 1.3 Seccion de EDA

Se realiza un pequeño analisis exploratorio de los datos para obsservar la distribucion del conjunto de datos


```python
def basic_eda(df, name):
    print(f"\n{'=' * 50}")
    print(f"EDA: {name}")
    print(f"{'=' * 50}")
    print(f"Dimensiones: {df.shape}")
    print("\nNulos:")
    print(df.isna().sum())
    print("\nCadenas vacías:")
    print(df[["en", "es"]].apply(lambda col: col.str.strip().eq("")).sum())
    print("\nPares duplicados:", df.duplicated(["en", "es"]).sum())
    print("\nEstadísticas de palabras:")
    for lang in ["en", "es"]:
        lengths = df[lang].str.split().str.len()
        print(f"\n{lang}:")
        print(lengths.describe(percentiles=[0.50, 0.75, 0.90, 0.95, 0.99]))
for name, df in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df),
]:
    basic_eda(df, name)
```

```python
def inspect_longest_pairs(df, lang="en", n=10):
    # Adaptamos los nombres a los que aparecen en tu salida: en y es.
    col = lang

    result = df.loc[
        df[col].str.split().str.len().nlargest(n).index
    ].copy()

    result["en_words"] = result["en"].str.split().str.len()
    result["es_words"] = result["es"].str.split().str.len()

    return result[
        ["en", "es", "en_words", "es_words"]
    ].sort_values("en_words", ascending=False)

inspect_longest_pairs(train_df, "en", 10)
```

```python
def add_quality_features(df):
    df = df.copy()

    for lang in ["en", "es"]:
        df[f"{lang}_words"] = df[lang].str.split().str.len()
        df[f"{lang}_chars"] = df[lang].str.len()

    df["word_ratio"] = (
        (df["en_words"] + 1) / (df["es_words"] + 1)
    )

    df["word_diff"] = (
        df["en_words"] - df["es_words"]
    ).abs()

    df["max_words"] = df[["en_words", "es_words"]].max(axis=1)

    return df


train_analysis = add_quality_features(train_df)


thresholds = [64, 128, 256, 512]

for limit in thresholds:
    mask = (
        (train_analysis["en_words"] > limit)
        | (train_analysis["es_words"] > limit)
    )

    count = int(mask.sum())

    print(
        f"> {limit:>3} palabras: "
        f"{count:>8,} pares "
        f"({count / len(train_analysis) * 100:.3f}%)"
    )
ratio_mask = (
    (train_analysis["word_ratio"] > 3)
    | (train_analysis["word_ratio"] < 1 / 3)
)

print("Pares con razón extrema:", ratio_mask.sum())
print(
    "Porcentaje:",
    f"{ratio_mask.mean() * 100:.3f}%"
)
```

Con este analisis previo, se encontró que existen varios elementos duplicados en el conjunto datos, sin embargo, el hallazgo mas interesante fue ver la diferencias fuertes de tokens que hay entre unos cuantos pares, pero a su vez, que es el problema mas fuerte, hay varias muestras que ocntiene una gran cantidad de tokens y se ve que son elementos muy largos o documentos.

# 3 Normalizaicon y tokenizacion
Para esta seccion, se intenta limpiar el conjunto de datos con tecncias como 
- Remover mayusculas
- Cuidar acentos
- Normalziar uNicode
- Reducir espacios
- Unificar apostrofes
La expresion regular propuesta busca o que haya apostrofes o que sean espacios para poder tokenizer, de esta manera, palabras como "don't", no se expanden. COn este regex, con la segunda alternativa despues del `|`, indica que o es una expresion u otra, y la segunda expresion es la que nos permite tokenizar a nivel palabra 

```python
import re
import unicodedata

TOKEN_PATTERN = re.compile(r"\w+(?:['’]\w+)*|[^\w\s]",flags=re.UNICODE,) 
def normalize_text(text):
    text = unicodedata.normalize("NFC", text)
    text = text.replace("’", "'")
    text = text.lower()
    text = re.sub(r"\s+", " ", text).strip()
    return text


def tokenize(text):
    normalized = normalize_text(text)
    return TOKEN_PATTERN.findall(normalized)
def detokenize(tokens):
    """Reconstruye texto a partir de tokens y ajusta la puntuación."""
    text = " ".join(tokens)
    # Quitar espacios antes de signos de puntuación.
    text = re.sub(r"\s+([.,!?;:%)\]}])", r"\1", text)
    # Quitar espacios después de signos de apertura.
    text = re.sub(r"([(\\[{¿¡])\s+", r"\1", text)
    # Reconstruir contracciones inglesas si aparecen.
    text = re.sub(r"\s+(['’])\s*", r"\1", text)
    return text.strip()
    
```

```python
def preprocess_corpus(df):
    result = df[["en", "es"]].copy()

    for lang in ("en", "es"):
        result[f"{lang}_norm"] = (
            result[lang].map(normalize_text)
        )
        result[f"{lang}_tokens"] = (
            result[f"{lang}_norm"].map(tokenize)
        )

    return result


train_processed = preprocess_corpus(train_clean_df)
validation_processed = preprocess_corpus(validation_clean_df)
test_processed = preprocess_corpus(test_cleaned_df)
```

# 3 Vocabulario
Se construye por medio de una clase, un vocabualrio, que contendra la frecuencia y la forma de acceder a cada token


```python
class Vocabulary:
    def __init__(self):
        self.token_freq= {}
        self.token_to_id= {}
        self.id_to_token={}
        self.unk_token = "<UNK>"
        self.pad_token = "<PAD>"
        self.unk_id = 0
        self.pad_id = 1
    def build(self,sentences):
        self.token_freq = {}
        for tokens in sentences:
            for token in tokens:
                self.token_freq[token]=(self.token_freq.get(token,0)+1)
        sorted_tokens=sorted(self.token_freq,key=lambda token:(-self.token_freq[token],token))
        special_tokens = [self.unk_token, self.pad_token]
        sorted_tokens = [token for token in sorted_tokens if token not in special_tokens]
        self.token_to_id= {
            token:idx
            for idx,token in enumerate(special_tokens)
        }
        for token in sorted_tokens:
            self.token_to_id[token] = len(self.token_to_id)
        self.id_to_token={
            idx:token
            for token,idx in self.token_to_id.items()
        }
    def __len__(self):
        return len(self.token_to_id)
    def get_id(self,token):
        return self.token_to_id.get(token,self.unk_id)
    def get_token(self,idx):
        return self.id_to_token.get(idx,self.unk_token)
    def get_frequency(self,token):
        return self.token_freq.get(token,0)
    def encode(self, tokens):
        return [self.get_id(token) for token in tokens]
    def decode(self, ids):
        return [self.get_token(token_id) for token_id in ids]
    def most_common(self,n=10):
        return sorted(self.token_freq.items(), key=lambda item:(-item[1],item[0]))[:n]
    def report(self):
        total_tokens = sum(self.token_freq.values())
        unique_tokens = len(self.token_freq)
        singletons = sum(
            1 for freq in self.token_freq.values()
            if freq == 1
        )

        return {
            "total_tokens": total_tokens,
            "unique_tokens": unique_tokens,
            "singletons": singletons,
            "singleton_ratio": (
                singletons / unique_tokens
                if unique_tokens else 0.0
            ),
        }
            
```

```python
en_vocab = Vocabulary()
es_vocab = Vocabulary()

en_vocab.build(train_processed["en_tokens"])
es_vocab.build(train_processed["es_tokens"])
```

```python
print("Tamaño del vocabulario inglés:", len(en_vocab))
print("Tamaño del vocabulario español:", len(es_vocab))

print("ID de 'the':", en_vocab.get_id("the"))
print("Token con ID 0:", en_vocab.get_token(0))
print("Frecuencia de 'the':", en_vocab.get_frequency("the"))

print("\nTokens más frecuentes en inglés:")
for token, frequency in en_vocab.most_common(10):
    print(f"{token!r}: {frequency:,}")

print("\nEstadísticas:")
print(en_vocab.report())

tokens = ["the", "cat", "supercalifragilistic"]

ids = en_vocab.encode(tokens)

print("Tokens:", tokens)
print("IDs:", ids)
print("Recuperación:", en_vocab.decode(ids))

```

```python
def unknown_report(vocab, df, lang):
    total = 0
    unknown = 0

    for tokens in df[f"{lang}_tokens"]:
        total += len(tokens)
        unknown += sum(
            token not in vocab.token_to_id
            for token in tokens
        )

    return {
        "total_tokens": total,
        "unknown_tokens": unknown,
        "unknown_rate": unknown / total if total else 0.0,
    }


print("Inglés - validación:", unknown_report(en_vocab, validation_processed, "en"))

print("Inglés - test:", unknown_report(en_vocab, test_processed, "en"))

print("Español - validación:", unknown_report(es_vocab, validation_processed, "es"))

print("Español - test:", unknown_report(es_vocab, test_processed, "es"))
```

# 4 Representacion vectorial del vocabulario
Ya que vamos a pasarlo por un conjunto de datos, tenemos que mandarlo a un mundo o representación vectorial, tanto puede ser como TFIDF u otra, en este caso y acorde a la literatura, se usan *matrices de coocurencia*, que lo que buscan es, si tengo una palabra en español ¿cuantas veces aparece en ingles?

Vamos a formular esta matriz de la siguiente forma
$$\mathcal{D}=\{(x^n,y^n)\}^N_{n=1}$$
Donde $x^n$ represent el vector de un idioma con sus tokens y el $y_n$, el otro par del vector. Y $N$, seria la cantidad total de filas que tiene nuestro dataset o mejor dicho, la cantidad de pares que tengamos en nuestro conjunto de datos.  Un ejemplo de esto seria 
| N | $x^n$ | $y^n$ |
|---|-------|-------|
|1| The cat is eating| El gato esta comiend| 
|---|-------|-------|

Como cada oracion, ya fue previamente tokenizada,entonces podemos reescribir cada oración como 
$$x^n=(x^n_1,\dots,x^n_{L_x})$$
$$y^n=(y^n_1,\dots,y^n_{L_y})$$
Donde el valor $L$, corresponde al valor máximo de tokens que tenga la frase, sin embargo, ahora que tenemos esta forma, ya tenemos los vocabularios de ambos idiomas como
$$\mathcal{V}_{en}=\{e_1,e_2,\dots,e_{|\mathcal{V}_{en}|}\}$$
$$\mathcal{V}_{es}=\{e_1,e_2,\dots,e_{|\mathcal{V}_{es}|}\}$$
Donde $e_1$, es cada palabra dentro del vocabulario y $|\mathcal{V}_{ex}|$ es la longitud de un determinado vocabulario. Estos tamaños finales, contemplan los caracteres especiales como `<UNK>` y `<PAD>`
## Definamos la matriz de coocurrencia

Ahora, es cuestión de como definir la matriz, y bajo la propuesta que definimos, diriamos que en una fila iria un idioma, y en las columnas iria el otro idioma, matematicamente esto se escribiria como

$$C \in \mathbb{N}^{\mathcal{V}_{en}\times \mathcal{V}_{es}}$$

Ahora, la idea es calcular la aparicion de una palabra en una frase, asi que 

$$f(e_i,x^n)=\begin{cases}
                1 & e_i \in x^n \\
                0 & e_i \notin x^n
\end{cases}$$

Esto mismo, pasaria para el otro vocabulario, de tal manera, que la matriz se construye como

$$C_{ij}=\sum^N_{n=1}f(en_i,x^n)f(es_j,y^n)$$

Finalmente, podemos obtener una probabildiad 
$$P(e_s|e_n)=\frac{C(e_n,e_s)}{\sum_{e'_s}C(e_n,e'_s)}$$

Con esta probabildiad, rpacticamente tenemos un mini traductor simple, que lo que dice es "que probabildiad tenemos de que un token en español sea dado a un token en ingles". Es decir, "que palabra posiblemente se asocia el ingles al español", y el calculo corresponde con la coocurrencia del token ingles y español entre la suma de ese token en ingles con todos los de español

```python
import math
import numpy as np
from scipy.sparse import csr_matrix

class BilingualCooccurrenceMatrix:
    def __init__(self, source_vocab, target_vocab):
        self.source_vocab = source_vocab
        self.target_vocab = target_vocab

        self.matrix = {}
        self.source_totals = {}
        self.target_totals = {}
        self.num_pairs = 0

    def fit(self, source_sentences, target_sentences):
        if len(source_sentences) != len(target_sentences):
            raise ValueError("Los corpus deben tener la misma cantidad de oraciones.")
        self.matrix = {}
        self.source_totals = {}
        self.target_totals = {}
        self.num_pairs = len(source_sentences)
        for source_tokens, target_tokens in zip(source_sentences, target_sentences):
            source_unique = set(source_tokens)
            target_unique = set(target_tokens)
            source_ids = {
                self.source_vocab.get_id(token)
                for token in source_unique
            }
            target_ids = {
                self.target_vocab.get_id(token)
                for token in target_unique
            }
            source_ids.discard(self.source_vocab.unk_id)
            source_ids.discard(self.source_vocab.pad_id)
            target_ids.discard(self.target_vocab.unk_id)
            target_ids.discard(self.target_vocab.pad_id)
            for source_id in source_ids:
                self.source_totals[source_id] = (self.source_totals.get(source_id, 0) + 1)
                row = self.matrix.setdefault(source_id, {})
                for target_id in target_ids:
                    row[target_id] = row.get(target_id, 0) + 1
            for target_id in target_ids:
                self.target_totals[target_id] = (self.target_totals.get(target_id, 0) + 1)
        return self

    def get_count(self, source_token, target_token):
        source_id = self.source_vocab.get_id(source_token)
        target_id = self.target_vocab.get_id(target_token)
        if source_id in (self.source_vocab.unk_id,self.source_vocab.pad_id,):
            return 0
        if target_id in (self.target_vocab.unk_id,self.target_vocab.pad_id):
            return 0

        return self.matrix.get(source_id, {}).get(target_id, 0)
    def get_translation_candidates(self, source_token, n=10):
        source_id = self.source_vocab.get_id(source_token)
        if source_id in (self.source_vocab.unk_id,self.source_vocab.pad_id,):
            return []
        row = self.matrix.get(source_id, {})
        candidates = sorted(row.items(),key=lambda item: (-item[1], item[0]),)[:n]
        return [
            (self.target_vocab.get_token(target_id), count)
            for target_id, count in candidates
        ]

    def get_probability(self, source_token, target_token):
        source_id = self.source_vocab.get_id(source_token)
        target_id = self.target_vocab.get_id(target_token)
        if source_id in (self.source_vocab.unk_id,self.source_vocab.pad_id,) or target_id in (self.target_vocab.unk_id,self.target_vocab.pad_id,):
            return 0.0
        row = self.matrix.get(source_id, {})
        total = sum(row.values())
        if total == 0:
            return 0.0
        return row.get(target_id, 0) / total

    def num_observed_pairs(self):
        return sum(len(row) for row in self.matrix.values())

    def get_pmi(self, source_token, target_token):
        source_id = self.source_vocab.get_id(source_token)
        target_id = self.target_vocab.get_id(target_token)
        if source_id in (self.source_vocab.unk_id,self.source_vocab.pad_id,) or target_id in (self.target_vocab.unk_id,self.target_vocab.pad_id,):
            return 0.0
        cooccurrence = self.matrix.get(source_id, {}).get(target_id, 0)
        source_count = self.source_totals.get(source_id, 0)
        target_count = self.target_totals.get(target_id, 0)
        if (cooccurrence == 0 or source_count == 0 or target_count == 0):
            return 0.0
        return math.log((cooccurrence * self.num_pairs)/ (source_count * target_count))
    def get_ppmi(self, source_token, target_token):
        return max(0.0, self.get_pmi(source_token, target_token))

    def get_ppmi_candidates(self, source_token, n=10, min_count=10):
        source_id = self.source_vocab.get_id(source_token)
        if source_id in (self.source_vocab.unk_id,self.source_vocab.pad_id,):
            return []
        row = self.matrix.get(source_id, {})
        candidates = []
        source_count = self.source_totals.get(source_id, 0)
        for target_id, count in row.items():
            if count < min_count:
                continue
            target_count = self.target_totals.get(target_id, 0)
            if source_count == 0 or target_count == 0:
                continue
            pmi = math.log((count * self.num_pairs)/ (source_count * target_count))
            ppmi = max(0.0, pmi)
            candidates.append((target_id, ppmi, count))
        candidates.sort(key=lambda item: (-item[1], item[0]))
        return [
            (self.target_vocab.get_token(target_id), score, count)
            for target_id, score, count in candidates[:n]
        ]

    def build_ppmi_matrix(self, min_count=200):
        source_ids = sorted(self.matrix.keys())
        target_ids = sorted(self.target_totals.keys())
        source_index = {
            token_id: i for i, token_id in enumerate(source_ids)
        }
        target_index = {
            token_id: i for i, token_id in enumerate(target_ids)
        }
        rows, cols, values = [], [], []
        N = self.num_pairs
        if N == 0:
            raise ValueError("La matriz no ha sido entrenada.")
        for source_id, row in self.matrix.items():
            source_count = self.source_totals.get(source_id, 0)
            if source_count == 0:
                continue
            for target_id, count in row.items():
                target_count = self.target_totals.get(target_id, 0)
                if count < min_count:
                    continue
                if count == 0 or target_count == 0:
                    continue
                pmi = np.log((count * N) / (source_count * target_count))
                if pmi > 0:
                    rows.append(source_index[source_id])
                    cols.append(target_index[target_id])
                    values.append(pmi)
        X = csr_matrix(
            (values, (rows, cols)),
            shape=(len(source_ids), len(target_ids)),
            dtype=np.float32,
        )

        return X, source_ids, target_ids
```

# 6 SVD proposicion

EN este caso, conviene utilizar SVD sobre PCA, debido a que estamos manejando matrices disperas, y para fines practicos y optimos, SVD es mas manejable a muy altas dimensionas con valores dispersos.
Sabiendo que ls filas representan las palabras en ingles y las columnas las palabras en español, entonces, la manera de obtener los elementos por medio de SVD, es tener en cuenta que al descomposición espectral que hace es 
$$X=U\Sigma V^T$$
Donde $U$, tiene las direcciones latentes de las palabras inglesas, la $V^T$, del español y $\Sigma$ los valores singulares ordenados del menor al mayor

A manera de obtener estas representacion,decimos que 
$$En=U\Sigma^{1/2}$$
$$Es=V\Sigma^{1/2}$$
De tal forma que ya no tomamos $n$, sino el valor $k$, de numero de componentes. Al hacer $1/2$, hacemos una representacion simetrica de la matriz de valores singulares y def esa manera, obtenemos ambos embeddings


```python
import numpy as np
from pathlib import Path
from sklearn.decomposition import TruncatedSVD
X_es_en, es_ids, en_ids = (cooc_matrix_es_en.build_ppmi_matrix(min_count=200))
print("Dimensiones de PPMI:", X_es_en.shape)
print("Asociaciones positivas:", X_es_en.nnz)
k = 1000
svd = TruncatedSVD(n_components=k,algorithm="randomized",n_iter=7,random_state=42)
es_embeddings = svd.fit_transform(X_es_en)
sqrt_s = np.sqrt(svd.singular_values_)
es_embeddings /= sqrt_s[None, :]
en_embeddings = svd.components_.T * sqrt_s[None, :]
print("Vectores españoles:", es_embeddings.shape)
print("Vectores ingleses:", en_embeddings.shape)
explained_variance = float(svd.explained_variance_ratio_.sum())
total_energy = np.sum(X_es_en.data.astype(np.float64) ** 2)
captured_energy = np.sum(svd.singular_values_.astype(np.float64) ** 2)
energy_ratio = (captured_energy / total_energy if total_energy > 0 else float("nan"))
print(f"Suma de varianza explicada: {explained_variance:.6f}")
print(f"Energía singular capturada: {energy_ratio:.6%}")
output_dir = Path("artifacts/embeddings_es_en")
output_dir.mkdir(parents=True, exist_ok=True)
np.save(output_dir / "es_embeddings.npy",es_embeddings.astype(np.float32))
np.save(output_dir / "en_embeddings.npy",en_embeddings.astype(np.float32))
np.savez_compressed(output_dir / "metadata.npz",es_ids=np.asarray(es_ids, dtype=np.int64),en_ids=np.asarray(en_ids, dtype=np.int64),
                    k=np.asarray(k),explained_variance=np.asarray(explained_variance),energy_ratio=np.asarray(energy_ratio))

print(f"Embeddings guardados en: {output_dir}")
```

## Evaluacion inicial de traduccion
Ahora, se va a evaluar el modelo de svd para observar que traducciones da y de ahi, poder implementar mejoras
Entonces, si incialmente teniamos $X_{ij}\in \mathbb{R}^{n \times m}$, siendo $n$, las palabras del vocablo ingles, mientras que $m$, del vocablo español y el $X_{ij}$ es el $PPMI$ de la palabra $e_{n_i}$ ingles y $e_{s_j}$ española. Asi que con las dimensiones que obtenemos, $PPMI \in \mathbb{R}^{185546 \times 238128}$ tras haber pasado por el SVD, recordemos que las matrices $U,\Sigma,V$,obtiene las siguientes dimensiones
$$
\begin{aligned}
U_k&\in\mathbb{R}^{n\times k}\\
\Sigma_k&\in\mathbb{R}^{k\times k}\\
V_k&\in\mathbb{R}^{m\times k}
\end{aligned}
$$
Donde $k$ es la reduccion que aplicamos, asi que tenemos los dos embeddings $En\in\mathbb{R}^{n\times k}$ y $Es\in\mathbb{R}^{m\times k}$, que representan el ingles y el español respectivamente y se obtiene apartir de la multiplicacion matricial simetrica entre $U$ y la matriz de eigenvalores y $V$ con la matriz de eigenvalores, esto anteriormente explicado.

Hasta el momento, es el resumen general que se hizo de operaciones realizadas despues de haber obtenido la matriz de coocurrencia y de haber calculado el svd

Asi que la pregunta interesante viene aqui y es, _¿Como traducimos?_
## Traduccion
Entonces, la traduccion se tiene que ver de la siguiente manera: teniendo los embeddings $En\in\mathbb{R}^{n\times k}$ y $Es\in\mathbb{R}^{m\times k}$. Podemos decir que $\mathbf{e} \in En$ y $\mathbf{s} \in Es$, donde cada vector entonces, se representa como $\mathbf{e},\mathbf{s}\in \mathbb{R}^{1 \times k}$. Y al tener ambos, misma dimension, podemos operar algebraicamente con ellos sin problemas, y tiene una interpretacion linguistica, ya quye estos vectores provienen de la matriz PPMI, asi que el producto punto $\mathbf{e}_i \cdot \mathbf{s}_j \approx X_{ij}$. que anteriormente estabamos usando. Adicionalmente, otra ventaja interesante, es que ya no solamente podemos utilizar el producto punto como medio de traduccion, sino que podemos emplear la distancia coseno
$$\cos(\mathbf{e}_i,\mathbf{s}_j)=\frac{\mathbf{e}_i \mathbf{s}^T_j }{\|\mathbf{e}_i\|_2 \|\mathbf{s}_j\|_2}$$
Asi que el planteamiento de la traduccion se convierte en: Dado un token en ingles, ¿cual es la probabilidad de que un token en español corresponda a este token?, buscandolo de la siguiente manera
$$s^*=\underset{s_j\in V_{es}}{\operatorname{argmax}}
\operatorname{sim}(\mathbf e,\mathbf s_j)
$$

Tambien se va a calcular la cantidad de veces que acierta el modelo con una metrica llamada HIT y MMR, en las siguiente nota se explicara en detalle estas metricas

```python
import numpy as np
from sklearn.preprocessing import normalize
def get_topk_translations(source_embedding,target_embedding,source_ids,target_ids,source_vocab,target_vocab,source_token,k=10):
    source_norm = normalize(source_embedding, norm="l2")
    target_norm = normalize(target_embedding, norm="l2")
    source_id_to_row = {
        token_id: i for i, token_id in enumerate(source_ids)
    }
    target_id_to_row = {
        token_id: i for i, token_id in enumerate(target_ids)
    }

    source_id = source_vocab.get_id(source_token)
    if source_id not in source_id_to_row:
        return []
    source_row = source_id_to_row[source_id]
    scores = target_norm @ source_norm[source_row]
    k = min(k, len(scores))
    if k == 0:
        return []
    top_indices = np.argpartition(scores, -k)[-k:]
    top_indices = top_indices[np.argsort(scores[top_indices])[::-1]]
    return [
        (target_vocab.get_token(target_ids[j]),float(scores[j]))
        for j in top_indices
    ]
```

```python
def evaluate_hit(gold_translations,source_embedding,target_embedding,source_ids,target_ids,source_vocab,target_vocab,ks=(1, 5, 10),):
    hits = {k: 0 for k in ks}
    evaluated = 0
    skipped = []
    for source_word, gold_words in gold_translations.items():
        source_id = source_vocab.get_id(source_word)
        if source_id not in set(source_ids):
            skipped.append(source_word)
            continue
        candidates = get_topk_translations(source_embedding,target_embedding,source_ids,target_ids,source_vocab,target_vocab,source_word,k=max(ks))
        print(f"\n{source_word} | Traducciones correctas: {gold_words}")
        for rank, (word, score) in enumerate(candidates, start=1):
            marker = " >" if word in gold_words else ""
            print(f"{rank:2}. {word:20} {score:.4f}{marker}")
        if not candidates:
            skipped.append(source_word)
            continue
        ranked_words = [word for word, score in candidates]
        evaluated += 1
        for k in ks:
            top_k_words = set(ranked_words[:k])
            if top_k_words.intersection(gold_words):
                hits[k] += 1
    if evaluated == 0:
        raise ValueError("No hay palabras disponibles para evaluar.")
    results = {
        f"Hit@{k}": hits[k] / evaluated
        for k in ks
    }
    return results, evaluated, skipped
```

### Métricas de evaluación: Hit@K y Mean Reciprocal Rank

Sea $Q$ el conjunto de consultas evaluables en ingles en este caso, $G(q)$ el conjunto de traducciones de referencia para una consulta, en este caso, en español. Definamos también $q$, como una palabra dentro del conjunto $Q$ de consulta y $R_K(q)$ la lista ordenada de las primeras $K$ traducciones predichas.

**Hit@K** mide la proporción de consultas en las que al menos una traducción de referencia aparece entre las primeras $K$ predicciones:

$$
\operatorname{Hit@K}
=\frac{1}{|Q|}\sum_{q\in Q}
\mathbb{1}\left[R_K(q)\cap G(q)\neq\varnothing\right].
$$

**Mean Reciprocal Rank (MRR@K)** mide el promedio del inverso del rango de la primera traducción correcta. Si no se encuentra una traducción de referencia entre las primeras $K$ predicciones, la contribución de la consulta es cero:

$$
\operatorname{MRR@K}
=\frac{1}{|Q|}\sum_{q\in Q}
\begin{cases}
1/r_q^*, & \text{si existe una coincidencia en Top-}K,\\
0, & \text{en otro caso},
\end{cases}
$$

donde $r_q^*$ es la posición de la primera traducción correcta.
$$
r_q^*=\min\{i\geq1:r_i\in G(q)\}
$$

Ambas métricas toman valores en $[0,1]$. Hit@K evalúa la presencia de una traducción correcta en las primeras posiciones, mientras que MRR@K incorpora información sobre su posición.

En el sistema propuesto, estas métricas se utilizan para comparar la búsqueda exhaustiva por similitud coseno con la búsqueda aproximada mediante Locality-Sensitive Hashing (LSH). Esta comparación permite estudiar el compromiso entre calidad de recuperación y eficiencia computacional.

## LSH
Este método, permite optimizar la busqueda, en vez de exhaustivamente, calcular una similitud coseno y ordenar con respecto a todos los valores de vectores para encontrar aquel vector más cercano con el que se busca la traducción. Este toma subconjuntos de vectores para poder hacer caclular los candidatos y sus similitudes coseno
$$N \rightarrow |C(q)| \rightarrow TopK(q)$$
Donde definiremos a $C(q)$, como el subconjunto de candidatos que toma LSH. Existe una variante del LSH, denominada _random hyperplane LSH_, donde
$$
h_a(x)=\begin{cases}
1 & \text{si} a^Tx\geq 0 \\
1 & \text{si} a^Tx< 0 
\end{cases}
$$
Donde $a \in \mathbb{R}^d$ es un vector normal con respecto a $x$ que indicara la dirección del vector. Con esto, podemos producir una firma en la región. Como tenemos 1000 dimensiones por el embedding y $12$ de los hiperplanos que aunque este valor es un hiperpárametro, inicialmente empezcamos con esto, entonces, las firmas están dadas por
$$z=x^TA$$
Donde $A\in \mathbb{R}^{d \times b}$ siendo $d=1000$ y $b=12$
Por lo tanto
$H(z)=(1[z_1\geq 0,\dots,z_b\geq 0])$
ESto implica, que cada vector podra tener 12 bits de firmas, implicando que puede tener $2^{12}=4096$ codigos diferentes.
Ahora, podemos definir la similitud coseno por una propiedad, en la que si tenemos dos vectores $x,y$, entonces la probabildidad de que un hiperplano se asigne al mismo bit para ambos vectores
$$p(s)=1-\frac{\arccos(s)}{\pi}$$
Donde s es $s=cos(\theta)$ y $\pi$, representa la normalizacion de la distancia coseno. Como se observa, el problema de asignacion termina siendo probaiblsitico, asi que para mejorar la precision del valor que pueda pertenecer a cierto hiperplanos, entonces hacemos un calculo de varias *tablas*.
Se proppone que la busqueda de probabildiad que cierta distancia coseno pertenezca cierto hiperplano se define como $p^b(s)$ de hiperplanos, pero para encontrar mayor exactitud, definimos $L$ tablas independientes que
$$
P_{\mathrm{recuperación}}
=1-(1-p_b)^L
$$

Abajo se evaluaran y comparan ambos metodos, el exhaustivo y LSH, a su vez, con se evaluan con las metricas propuestas.

```python
from time import perf_counter
import numpy as np
from sklearn.preprocessing import normalize
class TranslatorBilingual:
    DIRECTIONS = ("en-es", "es-en")
    def __init__(self, embeddings_input, vocabulary_es, vocabulary_en):
        self.vocabulary_es = vocabulary_es
        self.vocabulary_en = vocabulary_en
        self.en_embeddings = None
        self.es_embeddings = None
        self.metadata = None
        self.en_ids = None
        self.es_ids = None
        self.lsh_indices = {}
        self.load_embeddings(embeddings_input)
        self.prepare_embeddings()
    # ---------------------------------------------------------
    # 1. Carga y preparación de embeddings
    # ---------------------------------------------------------
    def load_embeddings(self, src):
        with np.load(f"{src}/metadata.npz",allow_pickle=False) as metadata:
            self.en_ids = metadata["en_ids"].copy()
            self.es_ids = metadata["es_ids"].copy()
        self.en_embeddings = np.load(f"{src}/en_embeddings.npy",allow_pickle=False)
        self.es_embeddings = np.load(f"{src}/es_embeddings.npy",allow_pickle=False)
    def prepare_embeddings(self):
        if (self.en_embeddings is None or self.es_embeddings is None or self.en_ids is None or self.es_ids is None ):
            raise ValueError("Los embeddings o sus metadatos no se han cargado.")
        if len(self.en_ids) != len(self.en_embeddings):
            raise ValueError("Los IDs ingleses no coinciden con los embeddings.")
        if len(self.es_ids) != len(self.es_embeddings):
            raise ValueError("Los IDs españoles no coinciden con los embeddings.")
        if self.en_embeddings.ndim != 2 or self.es_embeddings.ndim != 2:
            raise ValueError("Los embeddings deben ser matrices 2D.")
        if self.en_embeddings.shape[1] != self.es_embeddings.shape[1]:
            raise ValueError("Los embeddings deben tener la misma dimensionalidad.")
        if not np.isfinite(self.en_embeddings).all():
            raise ValueError("Los embeddings ingleses contienen valores no finitos.")
        if not np.isfinite(self.es_embeddings).all():
            raise ValueError("Los embeddings españoles contienen valores no finitos.")

        # Normalización L2 para calcular similitud coseno.
        self.en_norm = normalize(self.en_embeddings, norm="l2").astype(np.float32, copy=False)
        self.es_norm = normalize(self.es_embeddings, norm="l2").astype(np.float32, copy=False)
        # Relación entre ID del vocabulario y fila de la matriz.
        self.en_id_to_row = {
            int(token_id): i
            for i, token_id in enumerate(self.en_ids)
        }
        self.es_id_to_row = {
            int(token_id): i
            for i, token_id in enumerate(self.es_ids)
        }
        if len(self.en_id_to_row) != len(self.en_ids):
            raise ValueError("Hay IDs ingleses duplicados.")
        if len(self.es_id_to_row) != len(self.es_ids):
            raise ValueError("Hay IDs españoles duplicados.")
    # ---------------------------------------------------------
    # 2. Configuración de la dirección
    # ---------------------------------------------------------
    def _get_direction_data(self, direction):
        if direction == "en-es":
            return {
                "source_vocab": self.vocabulary_en,
                "target_vocab": self.vocabulary_es,
                "source_norm": self.en_norm,
                "target_norm": self.es_norm,
                "source_id_to_row": self.en_id_to_row,
                "target_ids": self.es_ids,
            }

        if direction == "es-en":
            return {
                "source_vocab": self.vocabulary_es,
                "target_vocab": self.vocabulary_en,
                "source_norm": self.es_norm,
                "target_norm": self.en_norm,
                "source_id_to_row": self.es_id_to_row,
                "target_ids": self.en_ids,
            }

        raise ValueError(
            "direction debe ser 'en-es' o 'es-en'."
        )

    def _get_source_row(self, token, direction):
        data = self._get_direction_data(direction)
        vocab = data["source_vocab"]
        token_id = vocab.get_id(token)
        if token_id in (vocab.unk_id, vocab.pad_id):
            return None
        return data["source_id_to_row"].get(int(token_id))
    # ---------------------------------------------------------
    # 3. Recuperación exhaustiva
    # ---------------------------------------------------------
    def get_topk_translations(self,token,k=10,direction="en-es"):
        if k <= 0:
            return []
        data = self._get_direction_data(direction)
        source_row = self._get_source_row(token, direction)
        if source_row is None:
            return []
        source_vector = data["source_norm"][source_row]
        target_norm = data["target_norm"]
        target_ids = data["target_ids"]
        target_vocab = data["target_vocab"]
        # Similitud coseno entre el origen y todos los destinos.
        scores = target_norm @ source_vector
        k = min(k, len(scores))
        if k == 0:
            return []
        top_indices = np.argpartition(scores, len(scores) - k)[-k:]
        top_indices = top_indices[np.argsort(scores[top_indices])[::-1]]
        return [(target_vocab.get_token(int(target_ids[j])),float(scores[j])) for j in top_indices]
    # ---------------------------------------------------------
    # 4. Hash LSH y construcción de índices
    # ---------------------------------------------------------
    @staticmethod
    def _lsh_hash(vector, planes):
        bits = vector @ planes >= 0
        code = 0
        for i, bit in enumerate(bits):
            if bit:
                code |= 1 << i
        return code

    def fit_lsh(self,direction="en-es",n_tables=10,n_planes=12,random_state=42):
        data = self._get_direction_data(direction)
        if not 1 <= n_planes <= 63:
            raise ValueError("n_planes debe estar entre 1 y 63.")
        if n_tables <= 0:
            raise ValueError("n_tables debe ser positivo.")
        target_norm = data["target_norm"]
        dimensions = target_norm.shape[1]
        rng = np.random.default_rng(random_state)
        planes_list = []
        tables_list = []
        for _ in range(n_tables):
            planes = rng.normal(size=(dimensions, n_planes)).astype(np.float32)
            table = {}
            for row, vector in enumerate(target_norm):
                code = self._lsh_hash(vector, planes)
                table.setdefault(code, []).append(row)
            planes_list.append(planes)
            tables_list.append(table)
        self.lsh_indices[direction] = {
            "planes": planes_list,
            "tables": tables_list,
            "n_tables": n_tables,
            "n_planes": n_planes,
        }

        return self

    def _get_lsh_candidate_rows(self, source_row, direction="en-es"):
        if direction not in self.lsh_indices:
            raise RuntimeError(f"Primero ejecuta fit_lsh(direction='{direction}').")
        data = self._get_direction_data(direction)
        index = self.lsh_indices[direction]
        query = data["source_norm"][source_row]
        candidates = set()
        for planes, table in zip(index["planes"], index["tables"]):
            code = self._lsh_hash(query, planes)
            candidates.update(table.get(code, []))
        return sorted(candidates)
    # ---------------------------------------------------------
    # 5. Recuperación LSH
    # ---------------------------------------------------------
    def get_topk_lsh(self,token,k=10,direction="en-es"):
        if k <= 0:
            return []
        if direction not in self.lsh_indices:
            raise RuntimeError(f"Primero ejecuta fit_lsh(direction='{direction}').")
        data = self._get_direction_data(direction)
        source_row = self._get_source_row(token, direction)
        if source_row is None:
            return []
        candidate_rows = self._get_lsh_candidate_rows(source_row, direction)
        if not candidate_rows:
            return []
        candidate_rows = np.asarray(candidate_rows, dtype=int)
        source_vector = data["source_norm"][source_row]
        target_norm = data["target_norm"]
        target_ids = data["target_ids"]
        target_vocab = data["target_vocab"]
        scores = target_norm[candidate_rows] @ source_vector
        k = min(k, len(scores))
        top_positions = np.argpartition(scores, len(scores) - k)[-k:]
        top_positions = top_positions[np.argsort(scores[top_positions])[::-1]]
        return [
            (target_vocab.get_token(int(target_ids[candidate_rows[pos]])),float(scores[pos]))
            for pos in top_positions
        ]
    # ---------------------------------------------------------
    # 6. Evaluación Hit@K y MRR
    # ---------------------------------------------------------
    def evaluate(self,gold_translations,k_values=(1, 5, 10),method="exhaustive",direction="en-es"):
        if not k_values or any(k <= 0 for k in k_values):
            raise ValueError("Todos los valores de K deben ser positivos.")
        if len(set(k_values)) != len(k_values):
            raise ValueError("No repitas valores de K.")
        if method not in ("exhaustive", "lsh"):
            raise ValueError("method debe ser 'exhaustive' o 'lsh'.")
        if method == "lsh" and direction not in self.lsh_indices:
            raise RuntimeError(f"Construye primero el índice LSH para '{direction}'.")
        self._get_direction_data(direction)
        hits = {k: 0 for k in k_values}
        reciprocal_ranks = []
        evaluated = 0
        skipped = []
        for source_word, gold in gold_translations.items():
            source_row = self._get_source_row(source_word, direction)
            if source_row is None:
                skipped.append(source_word)
                continue
            if isinstance(gold, str):
                gold = {gold}
            else:
                gold = set(gold)
            if not gold:
                skipped.append(source_word)
                continue
            if method == "exhaustive":
                predictions = self.get_topk_translations(source_word,k=max(k_values),direction=direction)
            else:
                predictions = self.get_topk_lsh(source_word,k=max(k_values),direction=direction)
            predicted_words = [word for word, _ in predictions ]
            evaluated += 1
            for k in k_values:
                if any(word in gold for word in predicted_words[:k] ):
                    hits[k] += 1
            rank = next((i for i, word in enumerate(predicted_words, start=1) if word in gold),None)
            reciprocal_ranks.append(1.0 / rank if rank is not None else 0.0)
        if evaluated == 0:
            raise ValueError("No se pudo evaluar ninguna palabra del conjunto.")
        return {
            **{
                f"hit@{k}": hits[k] / evaluated
                for k in k_values
            },
            f"mrr@{max(k_values)}": (
                sum(reciprocal_ranks) / evaluated
            ),
            "evaluated": evaluated,
            "skipped": skipped,
        }
    # ---------------------------------------------------------
    # 7. Comparación de recuperación exhaustiva y LSH
    # ---------------------------------------------------------
    def compare_retrievers(self,gold_translations,k_values=(1, 5, 10),direction="en-es",):
        if direction not in self.lsh_indices:
            raise RuntimeError(f"Ejecuta fit_lsh(direction='{direction}') antes de comparar.")
        data = self._get_direction_data(direction)
        valid_words = []
        skipped = []
        for word in gold_translations:
            if self._get_source_row(word, direction) is None:
                skipped.append(word)
            else:
                valid_words.append(word)
        if not valid_words:
            raise ValueError("No hay palabras válidas para realizar la comparación.")
        exhaustive_times = []
        lsh_times = []
        candidate_counts = []
        max_k = max(k_values)
        for word in valid_words:
            start = perf_counter()
            self.get_topk_translations(word, k=max_k, direction=direction)
            exhaustive_times.append(perf_counter() - start)
            source_row = self._get_source_row(word, direction)
            start = perf_counter()
            self.get_topk_lsh(word, k=max_k, direction=direction)
            lsh_times.append(perf_counter() - start)
            candidates = self._get_lsh_candidate_rows(source_row, direction)
            candidate_counts.append(len(candidates))
        exhaustive_metrics = self.evaluate(gold_translations,k_values=k_values,method="exhaustive",direction=direction)
        lsh_metrics = self.evaluate(gold_translations,k_values=k_values,method="lsh",direction=direction)
        exhaustive_ms = (np.mean(exhaustive_times) * 1000)
        lsh_ms = np.mean(lsh_times) * 1000
        mean_candidates = np.mean(candidate_counts)
        return {
            "direction": direction,
            "queries": len(valid_words),
            "skipped": skipped,
            "exhaustive": {
                "metrics": exhaustive_metrics,
                "mean_latency_ms": float(exhaustive_ms),
            },
            "lsh": {
                "metrics": lsh_metrics,
                "mean_latency_ms": float(lsh_ms),
                "mean_candidates": float(mean_candidates),
                "candidate_ratio": (
                    float(mean_candidates / len(data["target_ids"]))
                ),
            },
            "speedup": (
                float(exhaustive_ms / lsh_ms)
                if lsh_ms > 0
                else None
            ),
        }

    # ---------------------------------------------------------
    # 8. Traducción palabra por palabra
    # ---------------------------------------------------------
    def translate_sentence(self,sentence,k=1,direction="en-es",method="lsh",):
        if method not in ("lsh", "exhaustive"):
            raise ValueError("method debe ser 'lsh' o 'exhaustive'.")
        if method == "lsh" and direction not in self.lsh_indices:
            raise RuntimeError(f"Construye primero el índice LSH para '{direction}'.")
        source_tokens = tokenize(sentence)
        translated_tokens = []
        for token in source_tokens:
            # Conserva signos de puntuación.
            if not any(char.isalnum() for char in token):
                translated_tokens.append(token)
                continue
            if method == "lsh":
                candidates = self.get_topk_lsh(token, k=k, direction=direction)
            else:
                candidates = self.get_topk_translations(token, k=k, direction=direction)
            if candidates:
                translated_tokens.append(candidates[0][0])
            else:
                # Si no hay candidato, conserva el token original.
                translated_tokens.append(token)
        return detokenize(translated_tokens)
```

# Metricas finales

Para realizar una traducion a nivel de oracion, se necesitan ahora metricas como BLEU y chrf que permiten evalauir que tan bien tradujo una oracion
## BLEU
BLEU (Bilingual Evaluation Understudy) mide cuánto coincide la traducción generada con una o varias traducciones de referencia, por medio de los n grams, se calcula de la siguinete manera
$$
\operatorname{BLEU}
= BP \cdot \exp\left(
\sum_{n=1}^{N} w_n \log p_n
\right)
$$
Donde:
- $p_n$: precisión modificada de los n-gramas de tamaño $n$.
- $w_n$: peso asignado a cada tamaño; normalmente $1/4$ para n-gramas de 1 a 4 palabras.
- $BP$: penalización por brevedad (brevity penalty).
La penalización es:
$$
BP =
\begin{cases}
1 & c > r\\
e^{1-r/c} & 0 < c \leq r\\
0 & c=0
\end{cases}
$$   

Aquí $c$ es la longitud total de las traducciones generadas y $r$ la longitud de referencia correspondiente.
ESto se puede entender como que  BLEU premia que las palabras y sus secuencias coincidan con la referencia, pero puede penalizar paráfrasis válidas que usen otras palabras.

Por esta razón, en métricas de evaluación de traductores no solamente se cuenta métrica, sino también METEOR o chrf
# chrF: coincidencia de secuencias de caracteres
chrF (Character n-gram F-score) compara las traducciones a nivel de caracteres, en lugar de depender exclusivamente de palabras completas. Esto permite reconocer similitudes parciales, como palabras con la misma raíz, terminaciones similares o pequeñas diferencias ortográficas. Aquí si permite o es más flexible que BLEU. El como se calcula esta métrica es por medio de la obtención de n grams de la hipotesis y de la referencia . ESte calcula la precision y el recall a nivel caracter, y un $f_\beta$. Por lo general, la libreria calcula un $F_2$, dando mas penalizacion al presicion y potenciando el recall.

Como tal, no hay valores que indiquen que tan bien o mal, aunque estas métricas dependen del dataset, pero, por lo general si se obtienen valores muy cerca de 0, es que no hay muchas coincidencias, mientras que valores más arriba de 50-60 es que poseen muchas coincidencias
