## 1.2 Crear dataframe de analisis
Por facilidad, vamos a utilizar un dataframe de analisis para poder ser utilizado en los posteriores calculos

```python
import numpy
import pandas as pd 
def to_dataframe(split):
    translations=corpus[split]["translation"]
    return pd.DataFrame({
        "en":[pair["en"] for pair in translations],
        "es":[pair["es"] for pair in translations]
              })
train_df=to_dataframe("train")
val_df=to_dataframe("validation")
test_df=to_dataframe("test")
```

```python
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

|Validation | 2,000| 2|

## 1.3 Seccion de EDA

Se realiza un pequeño analisis exploratorio de los datos para obsservar la distribucion del conjunto de datos


```python
def basic_eda(df, name):
    print(f"\n{'=' * 50}")
    print(f"EDA: {name}")
    print(f"{'=' * 50}")
    print(f"Dimensiones: {df.shape}")
    print("\nNulos:")
    print(df.isna().sum())
    print("\nCadenas vacías:")
    print(df[["en", "es"]].apply(lambda col: col.str.strip().eq("")).sum())
    print("\nPares duplicados:", df.duplicated(["en", "es"]).sum())
    print("\nEstadísticas de palabras:")
    for lang in ["en", "es"]:
        lengths = df[lang].str.split().str.len()
        print(f"\n{lang}:")
        print(lengths.describe(percentiles=[0.50, 0.75, 0.90, 0.95, 0.99]))
for name, df in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df),
]:
    basic_eda(df, name)
```

```python
def inspect_longest_pairs(df, lang="en", n=10):
    # Adaptamos los nombres a los que aparecen en tu salida: en y es.
    col = lang

    result = df.loc[
        df[col].str.split().str.len().nlargest(n).index
    ].copy()

    result["en_words"] = result["en"].str.split().str.len()
    result["es_words"] = result["es"].str.split().str.len()

    return result[
        ["en", "es", "en_words", "es_words"]
    ].sort_values("en_words", ascending=False)

inspect_longest_pairs(train_df, "en", 10)
```

```python
def add_quality_features(df):
    df = df.copy()

    for lang in ["en", "es"]:
        df[f"{lang}_words"] = df[lang].str.split().str.len()
        df[f"{lang}_chars"] = df[lang].str.len()

    df["word_ratio"] = (
        (df["en_words"] + 1) / (df["es_words"] + 1)
    )

    df["word_diff"] = (
        df["en_words"] - df["es_words"]
    ).abs()

    df["max_words"] = df[["en_words", "es_words"]].max(axis=1)

    return df


train_analysis = add_quality_features(train_df)


thresholds = [64, 128, 256, 512]

for limit in thresholds:
    mask = (
        (train_analysis["en_words"] > limit)
        | (train_analysis["es_words"] > limit)
    )

    count = int(mask.sum())

    print(
        f"> {limit:>3} palabras: "
        f"{count:>8,} pares "
        f"({count / len(train_analysis) * 100:.3f}%)"
    )
ratio_mask = (
    (train_analysis["word_ratio"] > 3)
    | (train_analysis["word_ratio"] < 1 / 3)
)

print("Pares con razón extrema:", ratio_mask.sum())
print(
    "Porcentaje:",
    f"{ratio_mask.mean() * 100:.3f}%"
)
```

Con este analisis previo, se encontró que existen varios elementos duplicados en el conjunto datos, sin embargo, el hallazgo mas interesante fue ver la diferencias fuertes de tokens que hay entre unos cuantos pares, pero a su vez, que es el problema mas fuerte, hay varias muestras que ocntiene una gran cantidad de tokens y se ve que son elementos muy largos o documentos.

## 1.4 Limpieza del conjunto de datos 

EN la limpieza de conjunto de datos, se quitaron los documentos duplicado y posibles documentos que puedan ser un leakage para nuestros datos

```python
def duplicate_report(df, name):
    duplicated = df.duplicated(subset=["en", "es"])
    print(f"\n{name}")
    print(f"Total de pares: {len(df):,}")
    print(f"Filas duplicadas: {duplicated.sum():,}")
    print(f"Pares únicos: {df.drop_duplicates(['en', 'es']).shape[0]:,}")
    print(f"Porcentaje duplicado: {duplicated.mean() * 100:.3f}%")
def pair_set(df):
    return set(zip(df["en"], df["es"]))

    
for name, df in [("Train", train_df),("Validation", val_df),("Test", test_df)]:
    duplicate_report(df, name)


train_pairs = pair_set(train_df)
validation_pairs = pair_set(val_df)
test_pairs = pair_set(test_df)

print("\nTrain ∩ Validation:", len(train_pairs & validation_pairs))
print("Train ∩ Test:", len(train_pairs & test_pairs))
print("Validation ∩ Test:", len(validation_pairs & test_pairs))


###aplicamos la limpiea
train_clean_df=(train_df.drop_duplicates(subset=["en","es"]).reset_index(drop=True))
validation_clean_df=(val_df.drop_duplicates(subset=["en","es"]).reset_index(drop=True))
test_clean_df=(test_df.drop_duplicates(subset=["en","es"]).reset_index(drop=True))

validation_unique_pairs=set(zip(validation_clean_df["en"],validation_clean_df["es"]))
test_cleaned_df=(test_clean_df.loc[~test_clean_df.apply(lambda row: (row["en"],row["es"]) in validation_unique_pairs,axis=1 )].reset_index(drop=True))

## verficiacion de limpieaz
for name, df in [
    ("\nTrain", train_clean_df),
    ("Validation", validation_clean_df),
    ("Test", test_clean_df),
]:
    print(
        f"{name}: {len(df):,} filas | "
        f"{df.duplicated(['en', 'es']).sum():,} duplicados"
    )

train_pairs = set(zip(train_clean_df["en"], train_clean_df["es"]))
validation_pairs = set(zip(validation_clean_df["en"], validation_clean_df["es"]))
test_pairs = set(zip(test_cleaned_df["en"], test_clean_df["es"]))

print("Train ∩ Validation:", len(train_pairs & validation_pairs))
print("Train ∩ Test:", len(train_pairs & test_pairs))
print("Validation ∩ Test:", len(validation_pairs & test_pairs))
```

# 2 Corpus
Para esta parte es donde se comienza a realizar el desarrollo del formato de corpus para la representacion vectorial postuma. EL primer intento, es hacer que los conjuntos inicialmente cargados en memoria se guarden en un archivo

```python
import os

data_dir = os.path.join("data", "processed")
os.makedirs(data_dir, exist_ok=True)

corpora = {
    "train.csv": train_clean_df,
    "validation.csv": validation_clean_df,
    "test.csv": test_cleaned_df,
}

for filename, df in corpora.items():
    df[["en", "es"]].to_csv(os.path.join(data_dir, filename),index=False,encoding="utf-8",)
```
