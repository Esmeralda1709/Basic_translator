## 1.1 Lectura
Para el EDA inicial, se observara el tamaño, la existencia de caracteres nulos, duplicados exactos
La siguiente seccion, lee el conjunto de datos. En este caso, se muestra el conjunto de datos y una muestra

```python
from datasets import load_dataset
corpus= load_dataset("Helsinki-NLP/opus-100", "en-es")
```

```python
print(corpus)
print(type(corpus))
print(f"Lista de llaves{list(corpus.keys())}")
for split,data in corpus.items():
    print(f"{split}: {len(data):,} filas")
    print(type(data))
    print("Columnas:", data.column_names)
example = corpus["train"][0]

print("English:", example["translation"]["en"])
print("Spanish:", example["translation"]["es"])
```

Como se puede notar, el conjunto esta en un formato de json, y se interpreta como un diccionario de python, que tiene las claves de los splits definidos, como test,train y validation. EN cada uno tiene de columnas una traduccion y es un objeto de tipo arrow_dataset.Dataset. Esta es una facilidad que tiene la libreria `dataset` ya que nos permite descargar conjuntos de datos de hugging face y ejecutarlos de manera local
## 1.2 Crear dataframe de analisis
Por facilidad, vamos a utilizar un dataframe de analisis para poder ser utilizado en los posteriores calculos

```python
import numpy
import pandas as pd 
def to_dataframe(split):
    translations=corpus[split]["translation"]
    return pd.DataFrame({
        "en":translations["en"],
        "es":translations["es"]
    })
```
