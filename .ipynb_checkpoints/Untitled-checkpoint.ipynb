## 1.3 Seccion de EDA

Se realiza un pequeño analisis exploratorio de los datos para obsservar la distribucion del conjunto de datos


```python
def basic_eda(df, name):
    print(f"\n{'=' * 50}")
    print(f"EDA: {name}")
    print(f"{'=' * 50}")
    print(f"Dimensiones: {df.shape}")
    print("\nNulos:")
    print(df.isna().sum())
    print("\nCadenas vacías:")
    print(df[["en", "es"]].apply(lambda col: col.str.strip().eq("")).sum())
    print("\nPares duplicados:", df.duplicated(["en", "es"]).sum())
    print("\nEstadísticas de palabras:")
    for lang in ["en", "es"]:
        lengths = df[lang].str.split().str.len()
        print(f"\n{lang}:")
        print(lengths.describe(percentiles=[0.50, 0.75, 0.90, 0.95, 0.99]))
for name, df in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df),
]:
    basic_eda(df, name)
```

```python
def inspect_longest_pairs(df, lang="en", n=10):
    # Adaptamos los nombres a los que aparecen en tu salida: en y es.
    col = lang

    result = df.loc[
        df[col].str.split().str.len().nlargest(n).index
    ].copy()

    result["en_words"] = result["en"].str.split().str.len()
    result["es_words"] = result["es"].str.split().str.len()

    return result[
        ["en", "es", "en_words", "es_words"]
    ].sort_values("en_words", ascending=False)

inspect_longest_pairs(train_df, "en", 10)
```

```python
def add_quality_features(df):
    df = df.copy()

    for lang in ["en", "es"]:
        df[f"{lang}_words"] = df[lang].str.split().str.len()
        df[f"{lang}_chars"] = df[lang].str.len()

    df["word_ratio"] = (
        (df["en_words"] + 1) / (df["es_words"] + 1)
    )

    df["word_diff"] = (
        df["en_words"] - df["es_words"]
    ).abs()

    df["max_words"] = df[["en_words", "es_words"]].max(axis=1)

    return df


train_analysis = add_quality_features(train_df)


thresholds = [64, 128, 256, 512]

for limit in thresholds:
    mask = (
        (train_analysis["en_words"] > limit)
        | (train_analysis["es_words"] > limit)
    )

    count = int(mask.sum())

    print(
        f"> {limit:>3} palabras: "
        f"{count:>8,} pares "
        f"({count / len(train_analysis) * 100:.3f}%)"
    )
ratio_mask = (
    (train_analysis["word_ratio"] > 3)
    | (train_analysis["word_ratio"] < 1 / 3)
)

print("Pares con razón extrema:", ratio_mask.sum())
print(
    "Porcentaje:",
    f"{ratio_mask.mean() * 100:.3f}%"
)
```

Con este analisis previo, se encontró que existen varios elementos duplicados en el conjunto datos, sin embargo, el hallazgo mas interesante fue ver la diferencias fuertes de tokens que hay entre unos cuantos pares, pero a su vez, que es el problema mas fuerte, hay varias muestras que ocntiene una gran cantidad de tokens y se ve que son elementos muy largos o documentos.

## 1.4 Limpieza del conjunto de datos 

EN la limpieza de conjunto de datos, se quitaron los documentos duplicado y posibles documentos que puedan ser un leakage para nuestros datos

```python
def duplicate_report(df, name):
    duplicated = df.duplicated(subset=["en", "es"])
    print(f"\n{name}")
    print(f"Total de pares: {len(df):,}")
    print(f"Filas duplicadas: {duplicated.sum():,}")
    print(f"Pares únicos: {df.drop_duplicates(['en', 'es']).shape[0]:,}")
    print(f"Porcentaje duplicado: {duplicated.mean() * 100:.3f}%")
def pair_set(df):
    return set(zip(df["en"], df["es"]))

    
for name, df in [("Train", train_df),("Validation", val_df),("Test", test_df)]:
    duplicate_report(df, name)


train_pairs = pair_set(train_df)
validation_pairs = pair_set(val_df)
test_pairs = pair_set(test_df)

print("\nTrain ∩ Validation:", len(train_pairs & validation_pairs))
print("Train ∩ Test:", len(train_pairs & test_pairs))
print("Validation ∩ Test:", len(validation_pairs & test_pairs))


###aplicamos la limpiea
train_clean_df=(train_df.drop_duplicates(subset=["en","es"]).reset_index(drop=True))
validation_clean_df=(val_df.drop_duplicates(subset=["en","es"]).reset_index(drop=True))
test_clean_df=(test_df.drop_duplicates(subset=["en","es"]).reset_index(drop=True))

validation_unique_pairs=set(zip(validation_clean_df["en"],validation_clean_df["es"]))
test_cleaned_df=(test_clean_df.loc[~test_clean_df.apply(lambda row: (row["en"],row["es"]) in validation_unique_pairs,axis=1 )].reset_index(drop=True))

## verficiacion de limpieaz
for name, df in [
    ("\nTrain", train_clean_df),
    ("Validation", validation_clean_df),
    ("Test", test_clean_df),
]:
    print(
        f"{name}: {len(df):,} filas | "
        f"{df.duplicated(['en', 'es']).sum():,} duplicados"
    )

train_pairs = set(zip(train_clean_df["en"], train_clean_df["es"]))
validation_pairs = set(zip(validation_clean_df["en"], validation_clean_df["es"]))
test_pairs = set(zip(test_cleaned_df["en"], test_clean_df["es"]))

print("Train ∩ Validation:", len(train_pairs & validation_pairs))
print("Train ∩ Test:", len(train_pairs & test_pairs))
print("Validation ∩ Test:", len(validation_pairs & test_pairs))
```

# 3 Normalizaicon y tokenizacion
Para esta seccion, se intenta limpiar el conjunto de datos con tecncias como 
- Remover mayusculas
- Cuidar acentos
- Normalziar uNicode
- Reducir espacios
- Unificar apostrofes
La expresion regular propuesta busca o que haya apostrofes o que sean espacios para poder tokenizer, de esta manera, palabras como "don't", no se expanden. COn este regex, con la segunda alternativa despues del `|`, indica que o es una expresion u otra, y la segunda expresion es la que nos permite tokenizar a nivel palabra 

```python
import re
import unicodedata

TOKEN_PATTERN = re.compile(r"\w+(?:['’]\w+)*|[^\w\s]",flags=re.UNICODE,) 
def normalize_text(text):
    text = unicodedata.normalize("NFC", text)
    text = text.replace("’", "'")
    text = text.lower()
    text = re.sub(r"\s+", " ", text).strip()
    return text


def tokenize(text):
    normalized = normalize_text(text)
    return TOKEN_PATTERN.findall(normalized)
```

```python
def preprocess_corpus(df):
    result = df[["en", "es"]].copy()

    for lang in ("en", "es"):
        result[f"{lang}_norm"] = (
            result[lang].map(normalize_text)
        )
        result[f"{lang}_tokens"] = (
            result[f"{lang}_norm"].map(tokenize)
        )

    return result


train_processed = preprocess_corpus(train_clean_df)
validation_processed = preprocess_corpus(validation_clean_df)
test_processed = preprocess_corpus(test_cleaned_df)
```

# 3 Vocabulario
Se construye por medio de una clase, un vocabualrio, que contendra la frecuencia y la forma de acceder a cada token


```python
class Vocabulary:
    def __init__(self):
        self.token_freq= {}
        self.token_to_id= {}
        self.id_to_token={}
        self.unk_token = "<UNK>"
        self.pad_token = "<PAD>"
        self.unk_id = 0
        self.pad_id = 1
    def build(self,sentences):
        self.token_freq = {}
        for tokens in sentences:
            for token in tokens:
                self.token_freq[token]=(self.token_freq.get(token,0)+1)
        sorted_tokens=sorted(self.token_freq,key=lambda token:(-self.token_freq[token],token))
        special_tokens = [self.unk_token, self.pad_token]
        sorted_tokens = [token for token in sorted_tokens if token not in special_tokens]
        self.token_to_id= {
            token:idx
            for idx,token in enumerate(special_tokens)
        }
        for token in sorted_tokens:
            self.token_to_id[token] = len(self.token_to_id)
        self.id_to_token={
            idx:token
            for token,idx in self.token_to_id.items()
        }
    def __len__(self):
        return len(self.token_to_id)
    def get_id(self,token):
        return self.token_to_id.get(token,self.unk_id)
    def get_token(self,idx):
        return self.id_to_token.get(idx,self.unk_token)
    def get_frequency(self,token):
        return self.token_freq.get(token,0)
    def encode(self, tokens):
        return [self.get_id(token) for token in tokens]
    def decode(self, ids):
        return [self.get_token(token_id) for token_id in ids]
    def most_common(self,n=10):
        return sorted(self.token_freq.items(), key=lambda item:(-item[1],item[0]))[:n]
    def report(self):
        total_tokens = sum(self.token_freq.values())
        unique_tokens = len(self.token_freq)
        singletons = sum(
            1 for freq in self.token_freq.values()
            if freq == 1
        )

        return {
            "total_tokens": total_tokens,
            "unique_tokens": unique_tokens,
            "singletons": singletons,
            "singleton_ratio": (
                singletons / unique_tokens
                if unique_tokens else 0.0
            ),
        }
            
```

```python
en_vocab = Vocabulary()
es_vocab = Vocabulary()

en_vocab.build(train_processed["en_tokens"])
es_vocab.build(train_processed["es_tokens"])
```

```python
print("Tamaño del vocabulario inglés:", len(en_vocab))
print("Tamaño del vocabulario español:", len(es_vocab))

print("ID de 'the':", en_vocab.get_id("the"))
print("Token con ID 0:", en_vocab.get_token(0))
print("Frecuencia de 'the':", en_vocab.get_frequency("the"))

print("\nTokens más frecuentes en inglés:")
for token, frequency in en_vocab.most_common(10):
    print(f"{token!r}: {frequency:,}")

print("\nEstadísticas:")
print(en_vocab.report())

tokens = ["the", "cat", "supercalifragilistic"]

ids = en_vocab.encode(tokens)

print("Tokens:", tokens)
print("IDs:", ids)
print("Recuperación:", en_vocab.decode(ids))

```

```python
def unknown_report(vocab, df, lang):
    total = 0
    unknown = 0

    for tokens in df[f"{lang}_tokens"]:
        total += len(tokens)
        unknown += sum(
            token not in vocab.token_to_id
            for token in tokens
        )

    return {
        "total_tokens": total,
        "unknown_tokens": unknown,
        "unknown_rate": unknown / total if total else 0.0,
    }


print("Inglés - validación:", unknown_report(en_vocab, validation_processed, "en"))

print("Inglés - test:", unknown_report(en_vocab, test_processed, "en"))

print("Español - validación:", unknown_report(es_vocab, validation_processed, "es"))

print("Español - test:", unknown_report(es_vocab, test_processed, "es"))
```

# 4 Representacion vectorial del vocabulario
Ya que vamos a pasarlo por un conjunto de datos, tenemos que mandarlo a un mundo o representación vectorial, tanto puede ser como TFIDF u otra, en este caso y acorde a la literatura, se usan *matrices de coocurencia*, que lo que buscan es, si tengo una palabra en español ¿cuantas veces aparece en ingles?

Vamos a formular esta matriz de la siguiente forma
$$\mathcal{D}=\{(x^n,y^n)\}^N_{n=1}$$
Donde $x^n$ represent el vector de un idioma con sus tokens y el $y_n$, el otro par del vector. Y $N$, seria la cantidad total de filas que tiene nuestro dataset o mejor dicho, la cantidad de pares que tengamos en nuestro conjunto de datos.  Un ejemplo de esto seria 
| N | $x^n$ | $y^n$ |
|---|-------|-------|
|1| The cat is eating| El gato esta comiend| 
|---|-------|-------|

Como cada oracion, ya fue previamente tokenizada,entonces podemos reescribir cada oración como 
$$x^n=(x^n_1,\dots,x^n_{L_x})$$
$$y^n=(y^n_1,\dots,y^n_{L_y})$$
Donde el valor $L$, corresponde al valor máximo de tokens que tenga la frase, sin embargo, ahora que tenemos esta forma, ya tenemos los vocabularios de ambos idiomas como
$$\mathcal{V}_{en}=\{e_1,e_2,\dots,e_{|\mathcal{V}_{en}|}\}$$
$$\mathcal{V}_{es}=\{e_1,e_2,\dots,e_{|\mathcal{V}_{es}|}\}$$
Donde $e_1$, es cada palabra dentro del vocabulario y $|\mathcal{V}_{ex}|$ es la longitud de un determinado vocabulario. Estos tamaños finales, contemplan los caracteres especiales como `<UNK>` y `<PAD>`
## Definamos la matriz de coocurrencia

Ahora, es cuestión de como definir la matriz, y bajo la propuesta que definimos, diriamos que en una fila iria un idioma, y en las columnas iria el otro idioma, matematicamente esto se escribiria como

$$C \in \mathbb{N}^{\mathcal{V}_{en}\times \mathcal{V}_{es}}$$

Ahora, la idea es calcular la aparicion de una palabra en una frase, asi que 

$$f(e_i,x^n)=\begin{cases}
                1 & e_i \in x^n \\
                0 & e_i \notin x^n
\end{cases}$$

Esto mismo, pasaria para el otro vocabulario, de tal manera, que la matriz se construye como

$$C_{ij}=\sum^N_{n=1}f(en_i,x^n)f(es_j,y^n)$$

Finalmente, podemos obtener una probabildiad 
$$P(e_s|e_n)=\frac{C(e_n,e_s)}{\sum_{e'_s}C(e_n,e'_s)}$$

Con esta probabildiad, rpacticamente tenemos un mini traductor simple, que lo que dice es "que probabildiad tenemos de que un token en español sea dado a un token en ingles". Es decir, "que palabra posiblemente se asocia el ingles al español", y el calculo corresponde con la coocurrencia del token ingles y español entre la suma de ese token en ingles con todos los de español

```python
import math 
import numpy as np
from scipy.sparse import csr_matrix
class BilingualCooccurrenceMatrix:
    def __init__(self, en_vocab, es_vocab):
        self.en_vocab = en_vocab
        self.es_vocab = es_vocab
        self.matrix = {}
        self.en_totals = {}
        self.es_totals = {}
        self.num_pairs = 0

    def fit(self, english_sentences, spanish_sentences):
        if len(english_sentences) != len(spanish_sentences):
            raise ValueError("Los corpus deben tener la misma cantidad de oraciones.")
        self.matrix = {}
        self.en_totals = {}
        self.es_totals = {}
        self.num_pairs = len(english_sentences)
        for en_tokens, es_tokens in zip(english_sentences, spanish_sentences):
            en_unique = set(en_tokens)
            es_unique = set(es_tokens)
            en_ids = {self.en_vocab.get_id(token) for token in en_unique}
            es_ids = {self.es_vocab.get_id(token) for token in es_unique}
            en_ids.discard(self.en_vocab.unk_id)
            en_ids.discard(self.en_vocab.pad_id)
            es_ids.discard(self.es_vocab.unk_id)
            es_ids.discard(self.es_vocab.pad_id)

            for en_id in en_ids:
                self.en_totals[en_id] = (self.en_totals.get(en_id, 0) + 1)
                row = self.matrix.setdefault(en_id, {})
                for es_id in es_ids:
                    row[es_id] = row.get(es_id, 0) + 1
            for es_id in es_ids:
                self.es_totals[es_id] = (self.es_totals.get(es_id, 0) + 1)

    def get_count(self, en_token, es_token):
        en_id = self.en_vocab.get_id(en_token)
        es_id = self.es_vocab.get_id(es_token)

        if en_id == self.en_vocab.unk_id:
            return 0
        if es_id == self.es_vocab.unk_id:
            return 0

        return self.matrix.get(en_id, {}).get(es_id, 0)

    def get_translation_candidates(self, en_token, n=10):
        en_id = self.en_vocab.get_id(en_token)
        if en_id == self.en_vocab.unk_id:
            return []
        row = self.matrix.get(en_id, {})
        candidates = sorted(row.items(),key=lambda item: (-item[1], item[0]),)[:n]

        return [
            (self.es_vocab.get_token(es_id), count)
            for es_id, count in candidates
        ]

    def get_probability(self, en_token, es_token):
        en_id = self.en_vocab.get_id(en_token)
        es_id = self.es_vocab.get_id(es_token)

        if (en_id == self.en_vocab.unk_id or es_id == self.es_vocab.unk_id):
            return 0.0

        row = self.matrix.get(en_id, {})
        total = sum(row.values())

        if total == 0:
            return 0.0

        return row.get(es_id, 0) / total

    def num_observed_pairs(self):
        return sum(len(row) for row in self.matrix.values())
    def get_pmi(self,en_token,es_token):
        en_id = self.en_vocab.get_id(en_token)
        es_id = self.es_vocab.get_id(es_token)

        if en_id in (self.en_vocab.unk_id, self.en_vocab.pad_id):
            return 0.0

        if es_id in (self.es_vocab.unk_id, self.es_vocab.pad_id):
            return 0.0

        cooccurrence = self.matrix.get(en_id, {}).get(es_id, 0)
        en_count = self.en_totals.get(en_id, 0)
        es_count = self.es_totals.get(es_id, 0)

        if cooccurrence == 0 or en_count == 0 or es_count == 0:
            return 0.0

        return math.log(
            (cooccurrence * self.num_pairs)
            / (en_count * es_count)
        )
    def get_ppmi(self,en_token,es_token):
        return max(0,self.get_pmi(en_token,es_token))
        
    def get_ppmi_candidates(self, en_token, n=10,min_count=10):
        en_id = self.en_vocab.get_id(en_token)
        if en_id in (self.en_vocab.unk_id, self.en_vocab.pad_id):
            return []

        row = self.matrix.get(en_id, {})
        candidates=[]
        en_count = self.en_totals.get(en_id, 0)
        for es_id,count in row.items():
            if count < min_count:
                continue
            es_count = self.es_totals.get(es_id, 0)
            if en_count == 0 or es_count == 0:
                continue

            pmi = math.log(
                (count * self.num_pairs)
                / (en_count * es_count)
            )

            ppmi = max(0.0, pmi)

            candidates.append((es_id, ppmi, count))

        candidates.sort(key=lambda item: (-item[1], item[0]))

        return [(self.es_vocab.get_token(es_id), score, count) for es_id, score, count in candidates[:n]]
    def build_ppmi_matrix(self,min_count=200):
        en_ids = sorted(self.matrix.keys())
        es_ids = sorted(self.es_totals.keys())
        en_index = {token_id: i for i, token_id in enumerate(en_ids)}
        es_index = {token_id: i for i, token_id in enumerate(es_ids)}
        rows, cols, values = [], [], []
        N = self.num_pairs
        if N == 0:
            raise ValueError("La matriz no ha sido entrenada.")

        for en_id, row in self.matrix.items():
            en_count = self.en_totals.get(en_id, 0)
            if en_count == 0:
                continue
            for es_id, count in row.items():
                es_count = self.es_totals.get(es_id, 0)
                if count<min_count:
                    continue
                if count == 0 or es_count == 0:
                    continue

                pmi = np.log((count * N) / (en_count * es_count))

                if pmi > 0:
                    rows.append(en_index[en_id])
                    cols.append(es_index[es_id])
                    values.append(pmi)

        X = csr_matrix((values, (rows, cols)),shape=(len(en_ids), len(es_ids)),dtype=np.float32)

        return X, en_ids, es_ids
```

```python
cooc_matrix = BilingualCooccurrenceMatrix(en_vocab,es_vocab)

# Primero probamos con 1,000 pares.


cooc_matrix.fit(
    train_processed["en_tokens"],
    train_processed["es_tokens"],
)

print("Pares de oraciones:", cooc_matrix.num_pairs)
print("Asociaciones distintas:", cooc_matrix.num_observed_pairs())

print("Coocurrencia entre 'the' y 'el':",cooc_matrix.get_count("the", "el"),)

print("Candidatos para 'cat':",cooc_matrix.get_translation_candidates("cat", n=10),)
print(cooc_matrix.get_translation_candidates("cat", n=10))
print(cooc_matrix.get_ppmi_candidates("cat", n=10))
```

```python
print("Candidatos para 'house':",cooc_matrix.get_translation_candidates("house", n=10),)
print(cooc_matrix.get_ppmi_candidates("water", n=5,min_count=200))

gold_translations = {
    "cat": {"gato", "gata"},
    "dog": {"perro", "perra"},
    "house": {"casa"},
    "water": {"agua"},
    "book": {"libro"},
    "food": {"comida", "alimento"},
    "car": {"coche", "auto", "automóvil"},
    "city": {"ciudad"},
    "school": {"escuela"},
    "friend": {"amigo", "amiga"},
}

for en_word, gold_words in gold_translations.items():
    candidates = cooc_matrix.get_ppmi_candidates(
        en_word, n=10, min_count=200
    )

    ranked_words = [word for word, score, count in candidates]
    rank = next(
        (i + 1 for i, word in enumerate(ranked_words)
         if word in gold_words),
        None
    )

    print(f"{en_word:>8} | rango={rank} | {ranked_words[:5]}")
```

# Notas
Este metodo que fue, anteriormente aplicado, seria la primera linea base de traduccion, ya que sin necesidad de un modelo, compljeo, sino de probaibldiad bayesiana, podemos encontrar las traducciones de forma simple. Como se notara, la formulacion bayesiana se plantea de diferente manera ya que ahora no es querer intentar saber caul es la siguiente palabra o que la palabara ete asociada a cierta clase, sino que cierta palabra de un vocabulario, este asociado a cierta palabra de otro vocabulario.

Apartir de ahora, se probaran metodos alternos que intentaran optimizar la busqueda y uno de ellos, sera, generar una matriz de ppmi y reducir su dimensionalidad.

## SVD proposicion

EN este caso, conviene utilizar SVD sobre PCA, debido a que estamos manejando matrices disperas, y para fines practicos y optimos, SVD es mas manejable a muy altas dimensionas con valores dispersos.
Sabiendo que ls filas representan las palabras en ingles y las columnas las palabras en español, entonces, la manera de obtener los elementos por medio de SVD, es tener en cuenta que al descomposición espectral que hace es 
$$X=U\Sigma V^T$$
Donde $U$, tiene las direcciones latentes de las palabras inglesas, la $V^T$, del español y $\Sigma$ los valores singulares ordenados del menor al mayor

A manera de obtener estas representacion,decimos que 
$$En=U\Sigma^{1/2}$$
$$Es=V\Sigma^{1/2}$$
De tal forma que ya no tomamos $n$, sino el valor $k$, de numero de componentes. Al hacer $1/2$, hacemos una representacion simetrica de la matriz de valores singulares y def esa manera, obtenemos ambos embeddings


```python

import numpy as np
from pathlib import Path
from sklearn.decomposition import TruncatedSVD
X, en_ids, es_ids = cooc_matrix.build_ppmi_matrix(min_count=200)
print("Dimensiones de PPMI:", X.shape)
print("Asociaciones positivas:", X.nnz)
k = 1000
svd = TruncatedSVD(n_components=k,algorithm="randomized",n_iter=7,random_state=42,)
en_embeddings = svd.fit_transform(X)
sqrt_s = np.sqrt(svd.singular_values_)
en_embeddings /= sqrt_s[None, :]
es_embeddings = svd.components_.T * sqrt_s[None, :]
print("Vectores ingleses:", en_embeddings.shape)
print("Vectores españoles:", es_embeddings.shape)
explained_variance = float(svd.explained_variance_ratio_.sum())
total_energy = np.sum(X.data.astype(np.float64) ** 2)
captured_energy = np.sum(svd.singular_values_.astype(np.float64) ** 2)
energy_ratio = (captured_energy / total_energy if total_energy > 0else float("nan"))
print(f"Suma de varianza explicada: {explained_variance:.6f}")
print(f"Energía singular capturada: {energy_ratio:.6%}")
output_dir = Path("artifacts/embeddings")
output_dir.mkdir(parents=True, exist_ok=True)
np.save(output_dir / "en_embeddings.npy",en_embeddings.astype(np.float32),)
np.save(output_dir / "es_embeddings.npy",es_embeddings.astype(np.float32),)
np.savez_compressed(output_dir / "metadata.npz",en_ids=np.asarray(en_ids, dtype=np.int64),es_ids=np.asarray(es_ids, dtype=np.int64),
                    k=np.asarray(k),explained_variance=np.asarray(explained_variance),energy_ratio=np.asarray(energy_ratio))
print(f"Embeddings guardados en: {output_dir}")
```

## Evaluacion inicial de traduccion
Ahora, se va a evaluar el modelo de svd para observar que traducciones da y de ahi, poder implementar mejoras
Entonces, si incialmente teniamos $X_{ij}\in \mathbb{R}^{n \times m}$, siendo $n$, las palabras del vocablo ingles, mientras que $m$, del vocablo español y el $X_{ij}$ es el $PPMI$ de la palabra $e_{n_i}$ ingles y $e_{s_j}$ española. Asi que con las dimensiones que obtenemos, $PPMI \in \mathbb{R}^{185546 \times 238128}$ tras haber pasado por el SVD, recordemos que las matrices $U,\Sigma,V$,obtiene las siguientes dimensiones
$$
\begin{aligned}
U_k&\in\mathbb{R}^{n\times k}\\
\Sigma_k&\in\mathbb{R}^{k\times k}\\
V_k&\in\mathbb{R}^{m\times k}
\end{aligned}
$$
Donde $k$ es la reduccion que aplicamos, asi que tenemos los dos embeddings $En\in\mathbb{R}^{n\times k}$ y $Es\in\mathbb{R}^{m\times k}$, que representan el ingles y el español respectivamente y se obtiene apartir de la multiplicacion matricial simetrica entre $U$ y la matriz de eigenvalores y $V$ con la matriz de eigenvalores, esto anteriormente explicado.

Hasta el momento, es el resumen general que se hizo de operaciones realizadas despues de haber obtenido la matriz de coocurrencia y de haber calculado el svd

Asi que la pregunta interesante viene aqui y es, _¿Como traducimos?_
## Traduccion
Entonces, la traduccion se tiene que ver de la siguiente manera: teniendo los embeddings $En\in\mathbb{R}^{n\times k}$ y $Es\in\mathbb{R}^{m\times k}$. Podemos decir que $\mathbf{e} \in En$ y $\mathbf{s} \in Es$, donde cada vector entonces, se representa como $\mathbf{e},\mathbf{s}\in \mathbb{R}^{1 \times k}$. Y al tener ambos, misma dimension, podemos operar algebraicamente con ellos sin problemas, y tiene una interpretacion linguistica, ya quye estos vectores provienen de la matriz PPMI, asi que el producto punto $\mathbf{e}_i \cdot \mathbf{s}_j \approx X_{ij}$. que anteriormente estabamos usando. Adicionalmente, otra ventaja interesante, es que ya no solamente podemos utilizar el producto punto como medio de traduccion, sino que podemos emplear la distancia coseno
$$\cos(\mathbf{e}_i,\mathbf{s}_j)=\frac{\mathbf{e}_i \mathbf{s}^T_j }{\|\mathbf{e}_i\|_2 \|\mathbf{s}_j\|_2}$$
Asi que el planteamiento de la traduccion se convierte en: Dado un token en ingles, ¿cual es la probabilidad de que un token en español corresponda a este token?, buscandolo de la siguiente manera
$$s^*=\underset{s_j\in V_{es}}{\operatorname{argmax}}
\operatorname{sim}(\mathbf e,\mathbf s_j)
$$

Tambien se va a calcular la cantidad de veces que acierta el modelo con una metrica llamada HIT y MMR, en las siguiente nota se explicara en detalle estas metricas

```python
import numpy as np
from sklearn.preprocessing import normalize


def get_topk_translations(en_embedding,es_embedding,en_ids,es_ids,en_token, k=10):
    en_norm = normalize(en_embeddings, norm="l2")
    es_norm = normalize(es_embeddings, norm="l2")

    en_id_to_row = {token_id: i for i, token_id in enumerate(en_ids)}
    es_id_to_row = {token_id: i for i, token_id in enumerate(es_ids)}

    en_id = cooc_matrix.en_vocab.get_id(en_token)
    if en_id not in en_id_to_row:
        return []
    en_row = en_id_to_row[en_id]
    scores = es_norm @ en_norm[en_row]
    k = min(k, len(scores))
    if k == 0:
        return []
    top_indices = np.argpartition(scores, -k)[-k:]
    top_indices = top_indices[np.argsort(scores[top_indices])[::-1]]

    return [(cooc_matrix.es_vocab.get_token(es_ids[j]),float(scores[j]),)
        for j in top_indices]
```

```python

def evaluate_hit(gold_translations,en_embedding,es_embedding,en_ids,es_ids, ks=(1, 5, 10)):
    hits = {k: 0 for k in ks}
    evaluated = 0
    skipped = []

    for en_word, gold_words in gold_translations.items():
        en_id = cooc_matrix.en_vocab.get_id(en_word)
        if en_id not in en_id_to_row:
            skipped.append(en_word)
            continue
        candidates = get_topk_translations(en_embedding,es_embedding,en_ids,es_ids,en_word, k=max(ks))
        print(f"\n{en_word} | Traducciones correctas: {gold_words}")
        for rank, (word_, score) in enumerate(candidates, start=1):
            marker = " ✓" if word_ in gold_words else ""
            print(f"{rank:2}. {word_:20} {score:.4f}{marker}")
        if not candidates:
            skipped.append(en_word)
            continue

        # La función devuelve (palabra, similitud)
        ranked_words = [word for word, score in candidates]
        evaluated += 1
        for k in ks:
            top_k_words = set(ranked_words[:k])
            if top_k_words.intersection(gold_words):
                hits[k] += 1
    if evaluated == 0:
        raise ValueError("No hay palabras disponibles para evaluar.")
    results = {
        f"Hit@{k}": hits[k] / evaluated
        for k in ks
    }
    return results, evaluated, skipped


```

```python
for word in ["cat", "dog", "house", "water", "book"]:
    print(word, "→", get_topk_translations(en_embeddings,es_embeddings,en_ids,es_ids,word, k=5))
```

```python

gold_translations = {
    "cat": {"gato", "gata"},
    "dog": {"perro", "perra"},
    "house": {"casa"},
    "water": {"agua"},
    "book": {"libro"},
    "food": {"comida", "alimento"},
    "car": {"coche", "auto", "automóvil"},
    "city": {"ciudad"},
    "school": {"escuela"},
    "friend": {"amigo", "amiga"},
}


import numpy as np

# Cargar los embeddings guardados
en_embeddings = np.load("artifacts/embeddings/en_embeddings.npy")
es_embeddings = np.load("artifacts/embeddings/es_embeddings.npy")

metadata = np.load("artifacts/embeddings/metadata.npz")
en_ids = metadata["en_ids"]
es_ids = metadata["es_ids"]

hit_results, evaluated, skipped = evaluate_hit(gold_translations,en_embeddings,es_embeddings,en_ids,es_ids)

print(f"Palabras evaluadas: {evaluated}")
print(f"Palabras omitidas: {skipped}")

for metric, value in hit_results.items():
    print(f"{metric}: {value:.4f} ({value:.2%})")
```

# Este es nuestro primer modelo 
AHora vamos a definirlo en una buena clase, para poder ser reutilizable y poder optimizar algunos aprametros. Adicionalmente puede que convenga tambien gaurdar en cache el vocabulario


```python
from time import perf_counter
class TranslatorBilingual():
    def __init__(self,embeddings_input,vocabulary_es,vocabulary_en):
        self.en_embeddings = None
        self.es_embeddings = None
        self.metadata = None
        self.en_ids = None
        self.es_ids = None
        self.load_embeddings(embeddings_input)
        self.vocabulary_es=vocabulary_es
        self.vocabulary_en=vocabulary_en
        self.prepare_embeddings()
    def load_embeddings(self,src):
        self.en_embeddings = np.load(f"{src}/en_embeddings.npy")
        self.es_embeddings = np.load(f"{src}/es_embeddings.npy")
        self.metadata = np.load(f"{src}/metadata.npz")
        self.en_ids= self.metadata["en_ids"]
        self.es_ids = self.metadata["es_ids"]
    def prepare_embeddings(self):
        if self.en_embeddings is None or self.es_embeddings is None or self.metadata is None:
            raise ValueError("Esta corrupto o se necesitan cargar los embeddings")
        if len(self.en_ids) != len(self.en_embeddings):
            raise ValueError("Los IDs ingleses no coinciden con los embeddings.")
        if len(self.es_ids) != len(self.es_embeddings):
            raise ValueError("Los IDs españoles no coinciden con los embeddings.")
        self.en_norm = normalize(self.en_embeddings, norm="l2")
        self.es_norm = normalize(self.es_embeddings, norm="l2")
        self.en_id_to_row = {token_id: i for i, token_id in enumerate(self.en_ids)}
        self.es_id_to_row = {token_id: i for i, token_id in enumerate(self.es_ids)}
    def get_topk_translations(self,en_token,k=10):
        if k<=0:
            return []
        en_id=self.vocabulary_en.get_id(en_token)
        if en_id in(self.vocabulary_en.unk_id,self.vocabulary_en.pad_id):
            return []
        en_row=self.en_id_to_row[en_id]
        if en_row is None:
            return []
        scores= self.es_norm@self.en_norm[en_row]
        k = min(k, len(scores))
        if k == 0:
            return []
        top_indices = np.argpartition(scores, len(scores) - k)[-k:]
        top_indices = top_indices[np.argsort(scores[top_indices])[::-1]]
        return [(self.vocabulary_es.get_token(int(self.es_ids[j])),float(scores[j]))for j in top_indices]
    def evaluate(self,gold_translations,k_values=(1, 5, 10),method="exhaustive"):
        if not k_values or any(k <= 0 for k in k_values):
            raise ValueError("Todos los valores de K deben ser positivos.")
        if len(set(k_values)) != len(k_values):
            raise ValueError("No repitas valores de K.")
        max_k = max(k_values)
        hits = {k: 0 for k in k_values}
        reciprocal_ranks = []
        evaluated = 0
        skipped = []
    
        for en_word, gold in gold_translations.items():
            en_id = self.vocabulary_en.get_id(en_word)
            if (en_id in (self.vocabulary_en.unk_id,self.vocabulary_en.pad_id) or en_id not in self.en_id_to_row):
                skipped.append(en_word)
                continue
            if isinstance(gold, str):
                gold = {gold}
            else:
                gold = set(gold)
            if not gold:
                skipped.append(en_word)
                continue
            if method == "exhaustive":
                predictions = self.get_topk_translations(en_word, k=max_k)
            elif method == "lsh":
                predictions = self.get_topk_lsh(en_word, k=max_k)
            else:
                raise ValueError("method debe ser 'exhaustive' o 'lsh'.")
            predicted_words = [word for word, _ in predictions]
            evaluated += 1
            for k in k_values:
                if any(word in gold for word in predicted_words[:k]):
                    hits[k] += 1
            rank = next((i for i, word in enumerate(predicted_words, start=1) if word in gold),None,)
            reciprocal_ranks.append(1.0 / rank if rank is not None else 0.0)
        if evaluated == 0:
            raise ValueError("No se pudo evaluar ninguna palabra del conjunto.")
        return {
            **{f"hit@{k}": hits[k] / evaluated for k in k_values},
            f"mrr@{max_k}": sum(reciprocal_ranks) / evaluated,
            "evaluated": evaluated,
            "skipped": skipped,
        }
    def _lsh_hash(self, vector, planes):
        projections = vector @ planes
        bits = projections >= 0
    
        # Convertir los bits en un entero.
        code = 0
        for i, bit in enumerate(bits):
            if bit:
                code |= 1 << i
    
        return code


    def fit_lsh(self,n_tables=10,n_planes=12,random_state=42):
        if not 1 <= n_planes <= 63:
            raise ValueError("n_planes debe estar entre 1 y 63.")
    
        if n_tables <= 0:
            raise ValueError("n_tables debe ser positivo.")
    
        rng = np.random.default_rng(random_state)
        dimensions = self.es_norm.shape[1]
    
        self.lsh_planes = []
        self.lsh_tables = []
    
        for _ in range(n_tables):
            planes = rng.normal(
                size=(dimensions, n_planes)
            ).astype(np.float32)
    
            table = {}
    
            for row, vector in enumerate(self.es_norm):
                code = self._lsh_hash(vector, planes)
                table.setdefault(code, []).append(row)
    
            self.lsh_planes.append(planes)
            self.lsh_tables.append(table)
    
        return self


    def _get_lsh_candidate_rows(self, en_row):
        candidates = set()
        query = self.en_norm[en_row]  
        for planes, table in zip(self.lsh_planes, self.lsh_tables):
            code = self._lsh_hash(query, planes)
            candidates.update(table.get(code, []))
        return sorted(candidates)
    
    
    def get_topk_lsh(self, en_token, k=10):
        if k <= 0:
            return []
        if not hasattr(self, "lsh_tables"):
            raise RuntimeError("Primero debes construir el índice con fit_lsh().")
        en_id = self.vocabulary_en.get_id(en_token)
        if en_id in (self.vocabulary_en.unk_id,self.vocabulary_en.pad_id):
            return []
        en_row = self.en_id_to_row.get(en_id)
        if en_row is None:
            return []
        candidate_rows = self._get_lsh_candidate_rows(en_row)
        if not candidate_rows:
            return []
        candidate_rows = np.asarray(candidate_rows, dtype=int)
        scores = (self.es_norm[candidate_rows]@ self.en_norm[en_row])
        k = min(k, len(candidate_rows))
        top_positions = np.argpartition(scores, len(scores) - k)[-k:]
        top_positions = top_positions[np.argsort(scores[top_positions])[::-1]]
        return [
            (self.vocabulary_es.get_token(int(self.es_ids[candidate_rows[pos]])),float(scores[pos]))
            for pos in top_positions
        ]
    def compare_retrievers(self, gold_translations, k_values=(1, 5, 10)):
        if not hasattr(self, "lsh_tables"):
            raise RuntimeError("Debes ejecutar fit_lsh() antes de comparar.")
        if not k_values or any(k <= 0 for k in k_values):
            raise ValueError("Todos los valores de K deben ser positivos.")
        max_k = max(k_values)
        valid_words = []
        skipped = []
        for word in gold_translations:
            en_id = self.vocabulary_en.get_id(word)
            if (en_id in (self.vocabulary_en.unk_id,self.vocabulary_en.pad_id) or en_id not in self.en_id_to_row):
                skipped.append(word)
            else:
                valid_words.append(word)
        if not valid_words:
            raise ValueError("No hay palabras válidas para realizar la comparación.")
        timings = {
            "exhaustive": [],
            "lsh": [],
        }
        candidate_counts = []
        for word in valid_words:
            start = perf_counter()
            self.get_topk_translations(word, k=max_k)
            timings["exhaustive"].append(perf_counter() - start)
            en_id = self.vocabulary_en.get_id(word)
            en_row = self.en_id_to_row[en_id]
            start = perf_counter()
            self.get_topk_lsh(word, k=max_k)
            timings["lsh"].append(perf_counter() - start)
            candidates = self._get_lsh_candidate_rows(en_row)
            candidate_counts.append(len(candidates))
        exhaustive_metrics = self.evaluate(gold_translations,k_values=k_values,method="exhaustive",)
        lsh_metrics = self.evaluate(gold_translations,k_values=k_values,method="lsh",)
        exhaustive_ms = (sum(timings["exhaustive"]) / len(timings["exhaustive"])) * 1000
        lsh_ms = (sum(timings["lsh"]) / len(timings["lsh"])) * 1000
        return {
            "queries": len(valid_words),
            "skipped": skipped,
            "exhaustive": {
                "metrics": exhaustive_metrics,
                "mean_latency_ms": exhaustive_ms,
            },
            "lsh": {
                "metrics": lsh_metrics,
                "mean_latency_ms": lsh_ms,
                "mean_candidates": (sum(candidate_counts) / len(candidate_counts)),
                "candidate_ratio": (sum(candidate_counts)/ len(candidate_counts)/ len(self.es_ids)),
            },
            "speedup": (exhaustive_ms / lsh_ms if lsh_ms > 0 else None),
        }
```

```python
translator = TranslatorBilingual(
    embeddings_input="artifacts/embeddings",
    vocabulary_es=es_vocab,
    vocabulary_en=en_vocab,
)

print("Embeddings EN:", translator.en_embeddings.shape)
print("Embeddings ES:", translator.es_embeddings.shape)
print("IDs EN:", len(translator.en_ids))
print("IDs ES:", len(translator.es_ids))

assert translator.en_embeddings.shape[0] == len(translator.en_ids)
assert translator.es_embeddings.shape[0] == len(translator.es_ids)
assert translator.en_embeddings.shape[1] == translator.es_embeddings.shape[1]

print("Carga y dimensiones correctas.")
```

```python
for word in ["cat", "dog", "house", "water", "book"]:
    print(f"{word} -> {translator.get_topk_translations(word, k=5)}")
```
